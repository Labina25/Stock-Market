# Indian Stock Market Dataset – 5,258 Companies, P/E, Market Cap, ROCE & Growth


## Overview
This dataset contains stock market and financial information for 5,258 Indian companies. It brings together key market and fundamental indicators such as current stock price, P/E ratio, market capitalization, quarterly profit growth, quarterly sales growth, and Return on Capital Employed (ROCE).
The data is organized in a structured CSV format and can be used for stock market analysis, financial data analysis, exploratory data analysis, visualization, and machine learning projects.


## What's Inside
The dataset contains the following fields:

Company – Company name
Current_Price_INR – Current stock price in Indian Rupees (₹)
PE_Ratio – Price-to-Earnings ratio
Market_Cap_Crore – Market capitalization in ₹ crore
Quarterly_Profit_Growth_Percent – Quarterly profit growth percentage
Quarterly_Sales_Growth_Percent – Quarterly sales growth percentage
ROCE_Percent – Return on Capital Employed (ROCE)
Collection_Date – Date on which the data was collected

```python
# define numerical and categorical columns

numerical_features =[feature for feature in df.columns if df[feature].dtype!='O']
categorical_features =[feature for feature in df.columns if df[feature].dtype=='str']

# print columns
print('we have {} numerical features : {}'.format(len(numerical_features),numerical_features))
print('\n we have {} categorical features : {}'.format(len(categorical_features),categorical_features))
```

```python
num_cols =['Current_Price_INR', 'PE_Ratio', 'Market_Cap_Crore', 'Quarterly_Profit_Growth_Percent', 'Quarterly_Sales_Growth_Percent', 'ROCE_Percent', 'Expected_PE_Ratio']

for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df.info()
```

```python
## Add Columns for expected PE_Ratio
df['Expected_PE_Ratio'] = (df['Current_Price_INR'] / (1+ df['Quarterly_Profit_Growth_Percent'] /100))
```

# EDA : Exploratory data Analysis

```python
# Market cap segmentation
conditions = [
    (df['Market_Cap_Crore'] >= 20000),
    (df['Market_Cap_Crore'] >= 5000) & (df['Market_Cap_Crore'] < 20000),
    (df['Market_Cap_Crore'] >= 500) & (df['Market_Cap_Crore'] < 5000),
    (df['Market_Cap_Crore'] < 500),
]
labels =["Large Cap", "Mid Cap", "Small Cap", "Micro Cap"]
df["Cap_Category"] = np.select(conditions, labels, default="Unknown")

# Flag profitability
df['Is_profitable'] = (df['PE_Ratio'] > 0) & (df['PE_Ratio'].notnull())

df["Cap_Category"].value_counts().to_frame("Company Count")
```

```python
fig, axes = plt.subplots(1,2, figsize=(14,5))

# cap category count
sns.countplot(
    data=dff,
    x="Cap_Category",
    order =["Large Cap", "Mid Cap", "Small Cap", "Micro Cap"],
    ax= axes[0],
    palette="crest",

)
axes[0].set_title("Distribution Across Cap Categories")
axes[0].set_xlabel("")
axes[0].set_ylabel("Count")

## Log-transformed Market cap Distribution
sns.histplot(
    np.log10(df["Market_Cap_Crore"].dropna() +1),
    kde=True,
    bins=40,
    ax= axes[1],
    color='teal',
)

axes[1].set_title("Log10(Market Cap in cr)")
axes[1].set_xlabel("Log10(Market cap)")

plt.tight_layout()
plt.show()
```

```python
scatter_df = df[
    (df['PE_Ratio'] > 0)
    & (df['PE_Ratio'] < 100)
    & (df['ROCE_Percent'].between(-20,80))
]

plt.figure(figsize=(10,6))
sns.scatterplot(data = scatter_df, x="PE_Ratio", y="ROCE_Percent", hue="Cap_Category", alpha=0.6,
                palette="Dark2",)

plt.axhline(15, color="darkgreen",  label="ROCE =15%")
plt.axvline(25, color="red",  label="P/E =25%")

plt.title("ROCE vs P/E Ratio")
plt.xlabel("Price to earning (P/E)")
plt.ylabel("ROCE percent")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
corr_matrix = df[num_cols].corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="Blues",
    cbar=True,
    linewidths=0.5,
    square=True,
)

plt.title("Correlation Matrix of Financial Indecators")
plt.tight_layout()
plt.show()
```

```python
quality_screen = df[
    df["Cap_Category"].isin(["Large Cap", "Mid Cap"])
    & (df["ROCE_Percent"] > 20)
    & (df["PE_Ratio"].between(10,30))
    & (df["Quarterly_Sales_Growth_Percent"] > 10)

].sort_values(by="ROCE_Percent", ascending=False)

quality_screen[
    [
        'Company', 
        'Cap_Category',
        'Current_Price_INR',
        'PE_Ratio', 
        'ROCE_Percent',
        'Quarterly_Sales_Growth_Percent',    
    ]
].head(10).reset_index(drop=True)
```
